#### **CONCLUSIONES**

- **PRECION ACTUAL**: La media es de 1.030.501€ mientras que la mediana es de 620.000€. ESto se debe a que hay viviendas con precios muy altos que hacen que la media aumente y no represente correctamente el precio habitual de lso precios. Por ello es mas adecuado usar la *mediana*

- **PRECIO ANTERIOR**: La mayoria de valores de esta columna es 0, lo que significa que no disponemos de esta información y estos ceros deben tratarse como datos ausente, no como precios reales.

- **SUPERFICIE**: La mediana es de 103m^2 muetras que la media es de 154 m^2. De neuvo se ve afectada la media porque existen viviendas con superficies muy grandes. Hay valores que tienen que revisarse porque pueden ser errores, terrenos, edificios completos o promociones inmoviliarias. 

- **NUMERO DE HABITACIONES**: La mayoria de viviendas tiene alrededor de 3 habitaciones, si embargo existen anucios con hasta 20 habitaciones, por lo que hay que comprobar si se corresponden con viviendas reales de gran tamaño o a otro tipo de inmueble.

- **NUMERO DE BAÑOS**: Esta columna presenta un problema importante porque la mayoria de los anuncios tiene un valor de 0. No es razonable que tantas viviendas no tengan baño, por lo que lo más probable es que esta información no hay sido recogida correctamente. Como se ha comentado anteiormente, se busacra informacion complementaria en etiquetas, titulos y descripciones de lso anuncios.

- **ZONAS DE MADRID**: Este dataset contiene 21 zonas, que se ocrresponden con areas generales de Madrid. Sin embargo, el recomendador usara los 131 barrios oficiales para obtener resultados mucho mas concretos. Por tanto, mas adelante de identificara un barrio por anuncion a partir de su informacion.

- **ASCENSOR, LOCACLIZACION Y PLANTA**: La mayoría de las viviendas indican que tienen ascensor y que son exteriores.in embargo, estas columnas también contienen valores ausentes y diferentes formas de escribir la información, especialmente en el caso de la planta (1ª, BAJO, ENTREPLANTA, SÓTANO, valores negativos), por ello se transformara estos datos para que todos sigan el mismo formato.

- **ANUNCIOS DUPLICADOS**: No se han encontrado filas completamente repetidas ni enlaces duplicados, ya que cada anuncio tiene un enlace diferente. Aun así, podría ocurrir que una misma vivienda aparezca publicada varias veces con enlaces o descripciones diferentes, por lo que más adelante se hará una revisión más completa de posibles duplicados.

```python
print("Número de zonas:", df_raw["zona"].nunique())

display(
    df_raw["zona"]
    .value_counts(dropna=False)
    .rename_axis("zona")
    .reset_index(name="numero_anuncios")
)
```

En pirnciop el dataser ofrece cobertura de toda la ciudad. Sin embargo, el número de anuncios noe sta repartido de forma equilibrada.Solo Barrio de Salamanca y Centro reunen aproximadamente un tercio de todo los anuncios del dataset, en cambio las zonas con menos presencia son moratalaz y Barajas con solo 97 anuncios cada una. Esto significa que algunas zonas están mucho mejor representadas que otras y por tanto los anuncios de las zonas con muchos anuncios seran, en general, mas fiables que aquellas con una muestra pequeña.

```python
columnas_a_revisar = [
    "provincia",
    "ascensor",
    "localizacion",
    "planta",
    "habitaciones",
    "baños"
]

for columna in columnas_a_revisar:
    print("\n" + "=" * 60)
    print(f"COLUMNA: {columna}")
    print("=" * 60)

    resumen = (
        df_raw[columna]
        .value_counts(dropna=False)
        .rename_axis("valor")
        .reset_index(name="numero_anuncios")
    )

    resumen["porcentaje"] = (
        resumen["numero_anuncios"] / len(df_raw) * 100
    ).round(2)

    display(resumen)
```

La diferencia con la revisión anterior es que aquí se muestra la distribución completa y no solo los diez valores más frecuentes. Todavía no decide qué valores son incorrectos, solo los señala para revisarlos posteriormente.
